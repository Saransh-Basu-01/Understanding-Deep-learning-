```python
import numpy as np
import matplotlib.pyplot as plt

# 1D function (1 input)
def linear_function_1D(x, beta, omega):
    return beta + x * omega

# 2D function (2 inputs)
def linear_function_2D(x1, x2, beta, omega1, omega2):
    return beta + x1*omega1 + x2*omega2

# Create figure with 2 subplots
fig = plt.figure(figsize=(12, 5))

# PLOT 1: 1D function (2D graph: x vs y)
ax1 = fig.add_subplot(1, 2, 1)
x = np.arange(0, 10, 0.1)
y_1d = linear_function_1D(x, 10, -2)
ax1.plot(x, y_1d, 'r-', linewidth=2)
ax1.set_xlabel('x (input)')
ax1.set_ylabel('y (output)')
ax1.set_title('"1D" Function: y = f(x)\n(Plotted in 2D)')
ax1.grid(True, alpha=0.3)
ax1.axhline(y=0, color='black', alpha=0.2)
ax1.axvline(x=0, color='black', alpha=0.2)

# PLOT 2: 2D function (3D graph shown as contour)
ax2 = fig.add_subplot(1, 2, 2)
x1 = np.arange(0, 10, 0.1)
x2 = np.arange(0, 10, 0.1)
x1_mesh, x2_mesh = np.meshgrid(x1, x2)
y_2d = linear_function_2D(x1_mesh, x2_mesh, 0, 1, -0.5)
pos = ax2.contourf(x1_mesh, x2_mesh, y_2d, levels=256, cmap='hot', vmin=-10, vmax=10)
ax2.set_xlabel('x₁ (input 1)')
ax2.set_ylabel('x₂ (input 2)')
ax2.set_title('"2D" Function: y = f(x₁, x₂)\n(Shown as contour map)')
fig.colorbar(pos, ax=ax2, label='y (output)')

plt.tight_layout()
plt.show()

# Also show 3D surface view
fig = plt.figure(figsize=(8, 6))
ax3 = fig.add_subplot(111, projection='3d')
ax3.plot_surface(x1_mesh, x2_mesh, y_2d, cmap='hot', alpha=0.8)
ax3.set_xlabel('x₁')
ax3.set_ylabel('x₂')
ax3.set_zlabel('y')
ax3.set_title('"2D" Function in 3D view\n(2 inputs, 1 output)')
plt.show()
```

# A single linear equation with three inputs (i.e. linear_function_3D()) associates a value y with each point in a 3D space (x1,x2,x3). Is it possible to visualize this? What value is at position (0,0,0)?

## slicing method

```python
import numpy as np
import matplotlib.pyplot as plt

def linear_function_3D(x1, x2, x3, beta, omega1, omega2, omega3):
    return beta + x1*omega1 + x2*omega2 + x3*omega3

# Define parameters
beta = 5.0
omega1 = 2.0
omega2 = -1.0
omega3 = 0.5

# Create mesh for x1 and x2
x1 = np.arange(-10, 10, 0.2)
x2 = np.arange(-10, 10, 0.2)
x1_mesh, x2_mesh = np.meshgrid(x1, x2)

# Create 3 subplots for different x3 slices
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

x3_values = [-5, 0, 5]  # Different slices

for idx, x3_slice in enumerate(x3_values):
    # Fix x3 at a specific value
    y = linear_function_3D(x1_mesh, x2_mesh, x3_slice, beta, omega1, omega2, omega3)

    pos = axes[idx].contourf(x1_mesh, x2_mesh, y, levels=256, cmap='hot')
    axes[idx].set_title(f'x₃ = {x3_slice}')
    axes[idx].set_xlabel('x₁')
    axes[idx].set_ylabel('x₂')
    fig.colorbar(pos, ax=axes[idx])

plt.suptitle('Visualizing 3D Function by Slicing at Different x₃ Values')
plt.tight_layout()
plt.show()

print(f"Value at (0,0,0): {linear_function_3D(0, 0, 0, beta, omega1, omega2, omega3)}")
```

## 3D Surface with Color (Shows 4D)

```python
from mpl_toolkits.mplot3d import Axes3D

# Fix x3 at 0 and create 3D surface with color
x3_slice = 0
y = linear_function_3D(x1_mesh, x2_mesh, x3_slice, beta, omega1, omega2, omega3)

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')
surface = ax.plot_surface(x1_mesh, x2_mesh, y, cmap='hot', alpha=0.8, cstride=5, rstride=5)
ax.set_xlabel('x₁')
ax.set_ylabel('x₂')
ax.set_zlabel('y')
ax.set_title(f'3D Function (x₃ = {x3_slice})')
fig.colorbar(surface, label='y value')
plt.show()
```

```python
import numpy as np

def linear_function_2D(x1, x2, beta, omega1, omega2):
    return beta + x1*omega1 + x2*omega2

# Define parameters
beta1 = 0.5; beta2 = 0.2; beta3 = 0.3
omega11 = -1.0; omega12 = 0.4
omega21 = 0.1;  omega22 = 0.1
omega31 = 0.1;  omega32 = 0.4

# Define inputs
x1 = 4; x2 = -1

print("=" * 60)
print("INDIVIDUAL EQUATIONS")
print("=" * 60)

# Compute using individual equations
y1 = linear_function_2D(x1, x2, beta1, omega11, omega12)
y2 = linear_function_2D(x1, x2, beta2, omega21, omega22)
y3 = linear_function_2D(x1, x2, beta3, omega31, omega32)

print(f"y₁ = {beta1:.1f} + ({omega11:.1f})×{x1} + ({omega12:.1f})×{x2} = {y1:.3f}")
print(f"y₂ = {beta2:.1f} + ({omega21:.1f})×{x1} + ({omega22:.1f})×{x2} = {y2:.3f}")
print(f"y₃ = {beta3:.1f} + ({omega31:.1f})×{x1} + ({omega32:.1f})×{x2} = {y3:.3f}")

print("\n" + "=" * 60)
print("MATRIX/VECTOR FORM")
print("=" * 60)

# Define vectors and matrices
beta_vec = np.array([[beta1], [beta2], [beta3]])
omega_mat = np.array([[omega11, omega12],
                      [omega21, omega22],
                      [omega31, omega32]])
x_vec = np.array([[x1], [x2]])

print("Beta vector (β):")
print(beta_vec)
print("\nOmega matrix (W):")
print(omega_mat)
print("\nInput vector (x):")
print(x_vec)
print("\nMatrix multiplication: W × x")
print(np.matmul(omega_mat, x_vec))

# Compute with matrix form
y_vec = beta_vec + np.matmul(omega_mat, x_vec)

print("\nFinal result: β + W×x = y")
print("=" * 60)
print(f"y₁ = {y_vec[0][0]:.3f}")
print(f"y₂ = {y_vec[1][0]:.3f}")
print(f"y₃ = {y_vec[2][0]:.3f}")

# Verify both methods match
print("\n" + "=" * 60)
print("VERIFICATION")
print("=" * 60)
print(f"Individual vs Matrix form match: {np.allclose([y1, y2, y3], y_vec.flatten())}")

# Alternative using @ operator (Python 3.5+)
y_vec_alt = beta_vec + omega_mat @ x_vec
print(f"Using @ operator: {np.allclose(y_vec, y_vec_alt)}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 2, 100)
y = np.exp(x)

# Create points for chord demonstration
x1, x2 = -1.5, 1.5
y1, y2 = np.exp(x1), np.exp(x2)
t = 0.5  # Midpoint
x_mid = x1 * (1-t) + x2 * t
y_chord = y1 * (1-t) + y2 * t  # Chord value at midpoint
y_func = np.exp(x_mid)  # Function value at midpoint

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Plot 1: Basic exponential
ax = axes[0]
ax.plot(x, y, 'b-', linewidth=2, label='eˣ')
ax.axhline(y=0, color='black', alpha=0.3)
ax.axvline(x=0, color='black', alpha=0.3)
ax.grid(True, alpha=0.3)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Exponential Function eˣ')
ax.legend()
ax.set_ylim([-0.5, 8])

# Plot 2: Convexity demonstration
ax = axes[1]
x_plot = np.linspace(-2, 2, 100)
y_plot = np.exp(x_plot)

# Plot the function
ax.plot(x_plot, y_plot, 'b-', linewidth=2, label='eˣ')

# Draw chord between two points
x_points = [-1.5, 1.5]
y_points = [np.exp(-1.5), np.exp(1.5)]
ax.plot(x_points, y_points, 'r-', linewidth=2, label='Chord')

# Mark the points
ax.scatter(x_points, y_points, color='red', s=100, zorder=5)

# Show midpoint comparison
mid_x = 0
mid_y_func = np.exp(mid_x)
mid_y_chord = (y_points[0] + y_points[1]) / 2

ax.scatter([mid_x], [mid_y_func], color='green', s=100, zorder=5, label='f(midpoint)')
ax.scatter([mid_x], [mid_y_chord], color='orange', s=100, zorder=5, label='Chord at midpoint')

# Vertical line to show difference
ax.vlines(mid_x, mid_y_chord, mid_y_func, colors='purple', linestyles='dashed',
          label=f'f(x) < chord: {mid_y_func:.2f} < {mid_y_chord:.2f}')

ax.axhline(y=0, color='black', alpha=0.3)
ax.axvline(x=0, color='black', alpha=0.3)
ax.grid(True, alpha=0.3)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Convexity: Function below chord')
ax.legend()
ax.set_ylim([-0.5, 8])

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 2, 100)

fig, axes = plt.subplots(2, 1, figsize=(10, 8))

# Plot 1: Function and derivative
ax = axes[0]
ax.plot(x, np.exp(x), 'b-', linewidth=2, label='f(x) = eˣ')
ax.plot(x, np.exp(x), 'r--', linewidth=2, label="f'(x) = eˣ")
ax.axhline(y=0, color='black', alpha=0.3)
ax.axvline(x=0, color='black', alpha=0.3)
ax.grid(True, alpha=0.3)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Function and First Derivative')
ax.legend()

# Plot 2: Second derivative
ax = axes[1]
ax.plot(x, np.exp(x), 'g-', linewidth=2, label="f''(x) = eˣ > 0")
ax.axhline(y=0, color='black', alpha=0.3)
ax.axvline(x=0, color='black', alpha=0.3)
ax.grid(True, alpha=0.3)
ax.set_xlabel('x')
ax.set_ylabel("f''(x)")
ax.set_title('Second Derivative (Always Positive = Strictly Convex)')
ax.legend()
ax.fill_between(x, 0, np.exp(x), alpha=0.2, color='green')

plt.tight_layout()
plt.show()
```

```python
# Example: Why convexity matters for optimization
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 2, 100)
y = np.exp(x)

fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(x, y, 'b-', linewidth=2, label='eˣ (Convex)')

# Show gradient descent steps
x_start = 1.5
x_current = x_start
learning_rate = 0.2

for step in range(5):
    # Gradient is the derivative
    gradient = np.exp(x_current)
    x_next = x_current - learning_rate * gradient

    # Plot the steps
    ax.plot([x_current, x_current], [0, np.exp(x_current)], 'r--', alpha=0.5)
    ax.plot([x_current, x_next], [np.exp(x_current), np.exp(x_next)], 'ro-', alpha=0.5)

    # Annotate
    ax.text(x_current, np.exp(x_current) + 0.2, f'Step {step+1}', fontsize=9)

    x_current = x_next

ax.axhline(y=0, color='black', alpha=0.3)
ax.axvline(x=0, color='black', alpha=0.3)
ax.grid(True, alpha=0.3)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Gradient Descent on Convex Function eˣ')
ax.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Domain: x > 0
x = np.linspace(0.1, 5, 100)
y = np.log(x)

# Create points for chord demonstration
x1, x2 = 0.5, 4.0
y1, y2 = np.log(x1), np.log(x2)
t = 0.5  # Midpoint
x_mid = x1 * (1-t) + x2 * t
y_chord = y1 * (1-t) + y2 * t  # Chord value at midpoint
y_func = np.log(x_mid)  # Function value at midpoint

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Plot 1: Basic logarithm
ax = axes[0]
ax.plot(x, y, 'b-', linewidth=2, label='log(x)')
ax.axhline(y=0, color='black', alpha=0.3)
ax.axvline(x=0, color='black', alpha=0.3)
ax.grid(True, alpha=0.3)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Logarithm Function log(x)')
ax.legend()
ax.set_ylim([-3, 2])

# Plot 2: Concavity demonstration
ax = axes[1]
x_plot = np.linspace(0.1, 5, 100)
y_plot = np.log(x_plot)

# Plot the function
ax.plot(x_plot, y_plot, 'b-', linewidth=2, label='log(x)')

# Draw chord between two points
x_points = [0.5, 4.0]
y_points = [np.log(0.5), np.log(4.0)]
ax.plot(x_points, y_points, 'r-', linewidth=2, label='Chord')

# Mark the points
ax.scatter(x_points, y_points, color='red', s=100, zorder=5)

# Show midpoint comparison
mid_x = 2.25
mid_y_func = np.log(mid_x)
mid_y_chord = (y_points[0] + y_points[1]) / 2

ax.scatter([mid_x], [mid_y_func], color='green', s=100, zorder=5, label='f(midpoint)')
ax.scatter([mid_x], [mid_y_chord], color='orange', s=100, zorder=5, label='Chord at midpoint')

# Vertical line to show difference
ax.vlines(mid_x, mid_y_func, mid_y_chord, colors='purple', linestyles='dashed',
          label=f'f(x) > chord: {mid_y_func:.2f} > {mid_y_chord:.2f}')

ax.axhline(y=0, color='black', alpha=0.3)
ax.axvline(x=0, color='black', alpha=0.3)
ax.grid(True, alpha=0.3)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Concavity: Function above chord')
ax.legend()
ax.set_ylim([-3, 2])

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0.1, 5, 100)

fig, axes = plt.subplots(2, 1, figsize=(10, 8))

# Plot 1: Function and first derivative
ax = axes[0]
ax.plot(x, np.log(x), 'b-', linewidth=2, label='f(x) = log(x)')
ax.plot(x, 1/x, 'r--', linewidth=2, label="f'(x) = 1/x")
ax.axhline(y=0, color='black', alpha=0.3)
ax.axvline(x=0, color='black', alpha=0.3)
ax.grid(True, alpha=0.3)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Function and First Derivative')
ax.legend()
ax.set_ylim([-3, 5])

# Plot 2: Second derivative
ax = axes[1]
ax.plot(x, -1/x**2, 'g-', linewidth=2, label="f''(x) = -1/x² < 0")
ax.axhline(y=0, color='black', alpha=0.3)
ax.axvline(x=0, color='black', alpha=0.3)
ax.grid(True, alpha=0.3)
ax.set_xlabel('x')
ax.set_ylabel("f''(x)")
ax.set_title('Second Derivative (Always Negative = Strictly Concave)')
ax.legend()
ax.fill_between(x, -1/x**2, 0, alpha=0.2, color='red')
ax.set_ylim([-10, 1])

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# eˣ and its derivatives
x1 = np.linspace(-2, 2, 100)
ax = axes[0, 0]
ax.plot(x1, np.exp(x1), 'b-', linewidth=2, label='eˣ')
ax.plot(x1, np.exp(x1), 'r--', linewidth=2, label="(eˣ)' = eˣ")
ax.plot(x1, np.exp(x1), 'g-.', linewidth=2, label="(eˣ)'' = eˣ > 0")
ax.axhline(y=0, color='black', alpha=0.3)
ax.grid(True, alpha=0.3)
ax.set_title('eˣ: Second Derivative > 0 → CONVEX')
ax.legend()
ax.set_ylim([-0.5, 8])

# eˣ second derivative (positive)
ax = axes[0, 1]
ax.plot(x1, np.exp(x1), 'g-', linewidth=2)
ax.axhline(y=0, color='black', alpha=0.3)
ax.fill_between(x1, 0, np.exp(x1), alpha=0.3, color='green')
ax.set_title("f''(x) = eˣ > 0 (Convex)")
ax.set_ylim([-0.5, 8])

# log(x) and its derivatives
x2 = np.linspace(0.1, 5, 100)
ax = axes[1, 0]
ax.plot(x2, np.log(x2), 'b-', linewidth=2, label='log(x)')
ax.plot(x2, 1/x2, 'r--', linewidth=2, label="log'(x) = 1/x")
ax.plot(x2, -1/x2**2, 'g-.', linewidth=2, label="log''(x) = -1/x² < 0")
ax.axhline(y=0, color='black', alpha=0.3)
ax.axvline(x=0, color='black', alpha=0.3)
ax.grid(True, alpha=0.3)
ax.set_title('log(x): Second Derivative < 0 → CONCAVE')
ax.legend()
ax.set_ylim([-5, 5])

# log(x) second derivative (negative)
ax = axes[1, 1]
ax.plot(x2, -1/x2**2, 'r-', linewidth=2)
ax.axhline(y=0, color='black', alpha=0.3)
ax.fill_between(x2, -1/x2**2, 0, alpha=0.3, color='red')
ax.set_title("f''(x) = -1/x² < 0 (Concave)")
ax.set_ylim([-5, 1])

plt.tight_layout()
plt.show()
```

```python
# Example: Many probability distributions are log-concave
# This means their log is concave, making them nice to optimize

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

x = np.linspace(-4, 4, 100)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Normal distribution (log-concave)
ax = axes[0]
pdf = norm.pdf(x)
ax.plot(x, pdf, 'b-', linewidth=2, label='PDF (bell curve)')
ax.fill_between(x, pdf, alpha=0.3, color='blue')
ax.set_title('Normal Distribution (Log-Concave)')
ax.set_xlabel('x')
ax.set_ylabel('f(x)')
ax.grid(True, alpha=0.3)

# Log of normal distribution (concave)
ax = axes[1]
log_pdf = norm.logpdf(x)
ax.plot(x, log_pdf, 'r-', linewidth=2, label='log PDF (concave)')
ax.set_title('Log of Normal Distribution\n(Concave Function)')
ax.set_xlabel('x')
ax.set_ylabel('log(f(x))')
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Convex: Good for minimization
x = np.linspace(-2, 3, 100)
y = np.exp(x)

ax = axes[0]
ax.plot(x, y, 'b-', linewidth=2)
ax.plot([-1, 2], [np.exp(-1), np.exp(2)], 'r--', alpha=0.5)
ax.scatter([0.5], [np.exp(0.5)], color='green', s=100, zorder=5)
ax.axhline(y=0, color='black', alpha=0.3)
ax.axvline(x=0, color='black', alpha=0.3)
ax.grid(True, alpha=0.3)
ax.set_title('Convex: eˣ\nAny local min = global min')
ax.set_xlabel('x')
ax.set_ylabel('y')

# Show gradient descent
x_current = 1.5
for step in range(4):
    gradient = np.exp(x_current)
    x_next = x_current - 0.3 * gradient
    ax.plot([x_current, x_next], [np.exp(x_current), np.exp(x_next)], 'ro-', alpha=0.7)
    ax.text(x_current, np.exp(x_current)+0.2, f'Step {step+1}', fontsize=8)
    x_current = x_next

# Concave: Good for maximization
x = np.linspace(0.1, 4, 100)
y = np.log(x)

ax = axes[1]
ax.plot(x, y, 'r-', linewidth=2)
ax.plot([0.5, 3], [np.log(0.5), np.log(3)], 'b--', alpha=0.5)
ax.scatter([2], [np.log(2)], color='green', s=100, zorder=5)
ax.axhline(y=0, color='black', alpha=0.3)
ax.axvline(x=0, color='black', alpha=0.3)
ax.grid(True, alpha=0.3)
ax.set_title('Concave: log(x)\nAny local max = global max')
ax.set_xlabel('x')
ax.set_ylabel('y')

# Show gradient ascent
x_current = 0.5
for step in range(4):
    gradient = 1/x_current
    x_next = x_current + 0.3 * gradient  # Plus for ascent
    ax.plot([x_current, x_next], [np.log(x_current), np.log(x_next)], 'bo-', alpha=0.7)
    ax.text(x_current, np.log(x_current)+0.1, f'Step {step+1}', fontsize=8)
    x_current = x_next

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# 1. Minimizing convex function (eˣ)
ax = axes[0, 0]
x = np.linspace(-2, 2, 100)
y = np.exp(x)
ax.plot(x, y, 'b-', linewidth=2)
ax.axhline(y=0, color='black', alpha=0.3)
ax.set_title('Minimizing eˣ (Convex)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.grid(True, alpha=0.3)

# Show gradient descent
x_current = 1.5
for step in range(5):
    gradient = np.exp(x_current)
    x_next = x_current - 0.2 * gradient
    ax.plot([x_current, x_next], [np.exp(x_current), np.exp(x_next)], 'ro-')
    x_current = x_next
ax.scatter(x_current, np.exp(x_current), color='red', s=150, label='Minimum found')
ax.legend()

# 2. Maximizing concave function (log(x))
ax = axes[0, 1]
x = np.linspace(0.1, 4, 100)
y = np.log(x)
ax.plot(x, y, 'r-', linewidth=2)
ax.axhline(y=0, color='black', alpha=0.3)
ax.axvline(x=0, color='black', alpha=0.3)
ax.set_title('Maximizing log(x) (Concave)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.grid(True, alpha=0.3)

# Show gradient ascent
x_current = 0.5
for step in range(5):
    gradient = 1/x_current
    x_next = x_current + 0.3 * gradient
    ax.plot([x_current, x_next], [np.log(x_current), np.log(x_next)], 'bo-')
    x_current = x_next
ax.scatter(x_current, np.log(x_current), color='blue', s=150, label='Maximum found')
ax.legend()

# 3. What NOT to do: Maximizing convex
ax = axes[1, 0]
x = np.linspace(-2, 2, 100)
y = np.exp(x)
ax.plot(x, y, 'b-', linewidth=2)
ax.set_title('❌ Maximizing eˣ (Convex = BAD)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.grid(True, alpha=0.3)

# Gradient ascent on convex - diverges!
x_current = -1.5
for step in range(4):
    gradient = np.exp(x_current)
    x_next = x_current + 0.3 * gradient
    ax.plot([x_current, x_next], [np.exp(x_current), np.exp(x_next)], 'ro-')
    x_current = x_next
ax.text(0.5, 5, 'Goes to infinity!', color='red', fontsize=12, fontweight='bold')

# 4. What NOT to do: Minimizing concave
ax = axes[1, 1]
x = np.linspace(0.1, 4, 100)
y = np.log(x)
ax.plot(x, y, 'r-', linewidth=2)
ax.axhline(y=0, color='black', alpha=0.3)
ax.axvline(x=0, color='black', alpha=0.3)
ax.set_title('❌ Minimizing log(x) (Concave = BAD)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.grid(True, alpha=0.3)

# Gradient descent on concave - diverges!
x_current = 2.5
for step in range(4):
    gradient = 1/x_current
    x_next = x_current - 0.3 * gradient
    ax.plot([x_current, x_next], [np.log(x_current), np.log(x_next)], 'bo-')
    x_current = x_next
ax.text(0.5, -2, 'Goes to -∞!', color='blue', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()
```
